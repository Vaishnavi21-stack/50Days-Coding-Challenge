```python
students = {
    "Asha": {"Math": 78, "Science": 85, "English": 67},
    "Ravi": {"Math": 35, "Science": 40, "English": 50},
    "John": {"Math": 90, "Science": 92, "English": 88}
}

for name, marks in students.items():
    total = sum(marks.values())
    average = total / len(marks)

    print(f"{name} -> Total: {total}, Average: {average:.2f}")
```

```python
students = {
    "Asha": {"Math": 78, "Science": 85, "English": 67},
    "Ravi": {"Math": 35, "Science": 40, "English": 50},
    "John": {"Math": 90, "Science": 92, "English": 88}
}

top_scorer = ""
highest_total = 0

for name, marks in students.items():
    total = sum(marks.values())

    if total > highest_total:
        highest_total = total
        top_scorer = name

print("Top Scorer:", top_scorer)
print("Highest Total Marks:", highest_total)
```

```python
students = {
    "Asha": {"Math": 78, "Science": 85, "English": 67},
    "Ravi": {"Math": 35, "Science": 40, "English": 50},
    "John": {"Math": 90, "Science": 92, "English": 88}
}

failed_students = []

for name, marks in students.items():
    if any(mark < 40 for mark in marks.values()):
        failed_students.append(name)

print("Failed Students:", failed_students)
```
