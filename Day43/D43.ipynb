```python
import pandas as pd

df = pd.read_csv('Sample - Superstore.csv', encoding='latin1')

df.head()
```

```python
df.drop_duplicates(inplace=True)

df.duplicated().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x='Discount', y='Profit', data=df)

plt.title('Discount vs Profit')

plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

monthly_sales = df.groupby(df['Order Date'].dt.month)['Sales'].sum()

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')

plt.xlabel('Month')

plt.ylabel('Sales')

plt.show()
```

```python
region_profit = df.groupby('Region')['Profit'].sum()

sns.heatmap(region_profit.to_frame(), annot=True, cmap='coolwarm')

plt.title('Regional Profit Heatmap')

plt.show()
```

```python
loss_products = df.groupby('Sub-Category')['Profit'].sum().sort_values()

loss_products.plot(kind='barh', figsize=(8,6))

plt.title('Loss Making Products')

plt.xlabel('Profit')

plt.show()
```

```python
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

df['Delivery Days'] = (df['Ship Date'] - df['Order Date']).dt.days

delivery = df.groupby('Region')['Delivery Days'].mean()

delivery.plot(kind='bar')

plt.title('Average Delivery Days by Region')

plt.ylabel('Days')

plt.show()
```
