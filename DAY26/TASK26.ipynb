```python
students = ("Anu", "Ravikumar", "Jo")

longest_name = max(students, key=len)

print(longest_name)
```

```python
subjects = ("Math", "Science", "English")

result = ", ".join(subjects)

print(result)
```

```python
students = ("Anu", "Ravi")

masked = tuple(name[0] + "*" * (len(name) - 1) for name in students)

print(masked)
```

```python
students = ("Anu", "Ravi")

vowels = "aeiouAEIOU"

result = {name: sum(1 for ch in name if ch in vowels) for name in students}

print(result)
```

```python
courses = ("python", "sql", "power bi")

header = " | ".join(course.upper() for course in courses)

print(header)
```
