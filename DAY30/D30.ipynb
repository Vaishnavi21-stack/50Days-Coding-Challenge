```python
s = "abcabcbb"

char_set = set()
left = 0
max_length = 0

for right in range(len(s)):
    while s[right] in char_set:
        char_set.remove(s[left])
        left += 1

    char_set.add(s[right])
    max_length = max(max_length, right - left + 1)

print("Length of longest substring:", max_length)
```

```python
nums = [1, 2, 2, 3, 4, 3, 5]

unique_list = []
seen = set()

for num in nums:
    if num not in seen:
        unique_list.append(num)
        seen.add(num)

print("List after removing duplicates:", unique_list)
```

```python
numbers = (10, 5, 20, 8)

maximum = max(numbers)
minimum = min(numbers)

print("Maximum value:", maximum)
print("Minimum value:", minimum)
```

```python
text = "programming"

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print("Character Frequency:", frequency)
```

```python
words = ["cat", "dog", "elephant", "bat", "tiger"]

grouped = {}

for word in words:
    length = len(word)

    if length in grouped:
        grouped[length].append(word)
    else:
        grouped[length] = [word]

print("Grouped Words:", grouped)
```
