```python
data = "Ride101-Ramesh-Late Arrival-Bangalore, Ride102-Suresh-rude behavior-Hyderabad, Ride103-Mahesh-Late arrival-Bangalore"

clean_data = data.lower()

print(clean_data)
```

```python
complaints = clean_data.split(",")

total_complaints = len(complaints)

print(total_complaints)
```

```python
late_count = clean_data.count("late arrival")

print(late_count)
```

```python
bangalore_count = clean_data.count("bangalore")

print(bangalore_count)
```

```python
first_record = clean_data.split(",")[0]

driver_name = first_record.split("-")[1]

print(driver_name)
```

```python
standardized_data = clean_data.replace("rude behavior", "rude_behavior").replace("late arrival", "late_arrival")

print(standardized_data)
```
