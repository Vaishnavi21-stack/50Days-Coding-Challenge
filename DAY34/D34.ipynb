```python
ids = [101, 102, 101, 103, 102, 104]

unique_ids = []
seen = set()

for id in ids:
    if id not in seen:
        unique_ids.append(id)
        seen.add(id)

print(unique_ids)
```

```python
products = ["apple", "banana", "apple", "orange", "banana", "apple"]

frequency = {}

for item in products:
    frequency[item] = frequency.get(item, 0) + 1

print(frequency)
```

```python
marks = {"A": 85, "B": 92, "C": 88}

top_student = max(marks, key=marks.get)

print(top_student)
```

```python
skills1 = ["Python", "SQL", "Excel"]
skills2 = ["SQL", "Power BI", "Python"]

common_skills = list(set(skills1).intersection(set(skills2)))

print(common_skills)
```

```python
prices = {"apple": 50, "banana": 20, "orange": 30}

sorted_prices = sorted(prices.items(), key=lambda x: x[1])

print(sorted_prices)
```
