```python
nested = [[1, 2], [3, 4], [5]]

flattened = [item for sublist in nested for item in sublist]

print(flattened)
```

```python
def is_anagram(s1, s2):
    return sorted(s1) == sorted(s2)


# Test case
s1 = "listen"
s2 = "silent"

print(is_anagram(s1, s2))
```

```python
def find_missing(nums):
    full_set = set(range(1, max(nums) + 1))
    nums_set = set(nums)

    missing = full_set - nums_set
    return missing.pop()


# Test case
nums = [1, 2, 4, 5]
print(find_missing(nums))
```

```python
def group_words(words):
    result = {}

    for word in words:
        key = word[0]
        result.setdefault(key, []).append(word)

    return result


# Test case
words = ["apple", "ant", "banana", "bat"]
print(group_words(words))
```

```python
def remove_empty(data):
    return {k: v for k, v in data.items() if v is not None}


# Test case
data = {"a": 1, "b": None, "c": 3, "d": None}
print(remove_empty(data))
```
