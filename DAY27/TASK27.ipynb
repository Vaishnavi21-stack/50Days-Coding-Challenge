```python
trips = (
    (101, "Bangalore", 12.5, 250),
    (102, "Hyderabad", 8.0, 180),
    (103, "Bangalore", 15.2, 300),
    (104, "Chennai", 5.5, 120),
    (105, "Bangalore", 7.8, 200)
)
total_trips = len(trips)
total_trips
```

```python
bangalore_trips = sum(1 for trip in trips if trip[1] == "Bangalore")
bangalore_trips
```

```python
total_revenue = sum(trip[3] for trip in trips)
total_revenue
```

```python
high_distance_trips = [trip for trip in trips if trip[2] > 10]
high_distance_trips
```

```python
highest_fare_trip = max(trips, key=lambda trip: trip[3])
highest_fare_trip
```

```python
bangalore_fares = [trip[3] for trip in trips if trip[1] == "Bangalore"]
avg_bangalore_fare = sum(bangalore_fares) / len(bangalore_fares)
avg_bangalore_fare
```
