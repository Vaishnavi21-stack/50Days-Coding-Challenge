```python
batches = {}

for student in data:
    batch = student["batch"]

    if batch not in batches:
        batches[batch] = []

    batches[batch].append(student)

print(batches.keys())
```

```python
def calculate_average(values):
    return sum(values) / len(values)
```

```python
def batch_summary(batch_students):
    attendance = []
    consumption = []
    assignment = []

    for student in batch_students:
        attendance.append(student["live"])
        consumption.append(student["consumption"])
        assignment.append(student["assignment_sub"])

    return {
        "avg_attendance": calculate_average(attendance),
        "avg_consumption": calculate_average(consumption),
        "avg_assignment": calculate_average(assignment)
    }
```

```python
def find_at_risk_students(data):
    return [s for s in data if s["consumption"] < 20]


def find_top_students(data):
    return [s for s in data if s["assignment_score"] >= 80]
```

```python
def find_best_batch(batches):
    best_batch = None
    best_score = 0

    for name, students in batches.items():
        summary = batch_summary(students)

        score = (
            summary["avg_attendance"] +
            summary["avg_consumption"] +
            summary["avg_assignment"]
        ) / 3

        if score > best_score:
            best_score = score
            best_batch = name

    return best_batch
```

```python
print("Best Batch:", find_best_batch(batches))

print("\nBatch Summary:")
for name, students in batches.items():
    summary = batch_summary(students)
    print(f"{name} - Avg Consumption: {round(summary['avg_consumption'],1)}%, "
          f"Avg Attendance: {round(summary['avg_attendance'],1)}%")

print("\nAt-Risk Students (consumption < 20):")
for student in find_at_risk_students(data):
    print(student["name"], "-", student["batch"])
```
