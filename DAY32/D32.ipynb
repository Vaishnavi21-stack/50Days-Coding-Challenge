```python
# Even or Odd Checker

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# Positive, Negative or Zero Checker

num = float(input("Enter a number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

```python
# Print numbers from 1 to N using while loop

n = int(input("Enter a number: "))

i = 1

while i <= n:
    print(i)
    i += 1
```

```python
# Sum of first N numbers using while loop

n = int(input("Enter a number: "))

i = 1
total = 0

while i <= n:
    total += i
    i += 1

print("Sum =", total)
```

```python
# Password Retry System

correct_password = "admin123"
attempts = 0

while attempts < 3:
    password = input("Enter password: ")

    if password == correct_password:
        print("Access Granted")
        break
    else:
        print("Wrong password")
        attempts += 1

if attempts == 3:
    print("Access Denied")
```

```python
# Multiplication Table

num = int(input("Enter a number: "))

i = 1

while i <= 10:
    print(f"{num} x {i} = {num * i}")
    i += 1
```
