```python
students = ("sunitha", "RAVI", "aNu")
formatted_students = tuple(name.capitalize() for name in students)
print(formatted_students)
```

```python
subjects = ("Math", "Science", "English", "Math", "Science")

total_subjects = len(subjects)
unique_subjects = set(subjects)

print("Total Subjects:", total_subjects)
print("Unique Subjects:", unique_subjects)
```

```python
feedback = ("  good course  ", " excellent ", "average  ")
clean_feedback = tuple(text.strip() for text in feedback)
print(clean_feedback)
```

```python
names = ("Anu", "Ravi", "John")
courses = ("Python", "SQL", "Power BI")
combined = tuple(name + "-" + course for name, course in zip(names, courses))
print(combined)
```

```python
emails = ("anu@gmail.com", "ravi@yahoo.com", "john@edu.com")
domains = tuple(email.split("@")[1] for email in emails)
print(domains)
```
