```python
import pandas as pd

data = [
["Bangalore South","Karnataka","ABC","Ravi Kumar",45231,120000,"Male",45],
["Bangalore South","Karnataka","XYZ","Sneha Reddy",50321,120000,"Female",38],
["Bangalore South","Karnataka","PQR","Amit Shah",18448,120000,"Male",52],
["Mysore","Karnataka","ABC","Kiran Gowda",62341,150000,"Male",41],
["Mysore","Karnataka","XYZ","Divya Sharma",54321,150000,"Female",36],
["Mysore","Karnataka","PQR","Rahul Verma",33211,150000,"Male",48],
["Tumkur","Karnataka","ABC","Suresh Naik",41234,110000,"Male",50],
["Tumkur","Karnataka","XYZ","Anita Desai",45210,110000,"Female",39],
["Tumkur","Karnataka","PQR","Manoj Singh",23556,110000,"Male",44]
]

df = pd.DataFrame(data, columns=[
"Constituency","State","Party","Candidate","Votes","Total_Votes","Gender","Age"
])

df.head()
```

```python
# Check duplicate rows
duplicates = df.duplicated()

print("Number of duplicate rows:", duplicates.sum())

# Show duplicates if any
df[duplicates]
```

```python
df = df.drop_duplicates()
```

```python
df.dtypes
```

```python
df["Votes"] = pd.to_numeric(df["Votes"], errors="coerce")
df["Total_Votes"] = pd.to_numeric(df["Total_Votes"], errors="coerce")
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
df.dtypes
```

```python
df.groupby("Constituency")["Total_Votes"].nunique()
```

```python
df.groupby("Constituency")["Total_Votes"].nunique().reset_index()
```

```python
df["Vote %"] = (df["Votes"] / df["Total_Votes"]) * 100
```

```python
df.head()
```

```python
df["Vote %"] = df["Vote %"].round(2)
```

```python
df[["Candidate", "Votes", "Total_Votes", "Vote %"]].head()
```

```python
# Get index of max votes per constituency
idx = df.groupby("Constituency")["Votes"].idxmax()

# Extract winner rows
winners = df.loc[idx][["Constituency", "Candidate", "Party", "Votes"]]

winners
```

```python
# Sort votes within each constituency
df_sorted = df.sort_values(["Constituency", "Votes"], ascending=[True, False])

# Get top 2 candidates per constituency
top2 = df_sorted.groupby("Constituency").head(2)

# Calculate margin
margin_df = top2.groupby("Constituency").agg({
    "Votes": lambda x: x.iloc[0] - x.iloc[1]
}).reset_index()

margin_df.rename(columns={"Votes": "Winning Margin"}, inplace=True)

margin_df
```

```python
winners = winners.merge(margin_df, on="Constituency")
winners
```

```python
# Sort votes descending within each constituency
df_sorted = df.sort_values(["Constituency", "Votes"], ascending=[True, False])

# Get top 2 candidates per constituency
top2 = df_sorted.groupby("Constituency").head(2)

# Calculate winning margin
margin_df = top2.groupby("Constituency").agg({
    "Votes": lambda x: x.iloc[0] - x.iloc[1]
}).reset_index()

margin_df.rename(columns={"Votes": "Winning Margin"}, inplace=True)

margin_df.head()
```

```python
# Recreate winners if not present
idx = df.groupby("Constituency")["Votes"].idxmax()
winners = df.loc[idx]

# Merge margin
winners = winners.merge(margin_df, on="Constituency")

winners[["Constituency","Candidate","Party","Votes","Winning Margin"]]
```

```python
close_contests = winners[winners["Winning Margin"] < 5000]

close_contests[["Constituency","Candidate","Winning Margin"]]
```
